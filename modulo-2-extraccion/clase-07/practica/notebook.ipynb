### Ejercicio 7 — JOIN desde Python
 
La tabla `mascotas` tiene el `dueno_id`, pero no el nombre del dueño. Para verlos juntos necesitás unir las dos tablas, igual que hacías en SQLBolt.
 
**Consigna:** Escribe una consulta con `JOIN` para traer el nombre de la mascota, su especie y el nombre de su dueño, todo en un solo DataFrame. No uses dos `pd.read_sql()` separados y luego los combines — hacelo en una sola consulta SQL.

### Ejercicio 8 — Dos JOINs encadenados
 
Para ver el nombre del dueño junto con los datos de cada consulta médica, necesitas pasar por tres tablas: `consultas` → `mascotas` → `duenos`.
 
**Consigna:** Escribe una consulta que una las tres tablas y devuelva: nombre del dueño, nombre de la mascota, fecha de la consulta, motivo y costo. Ordená el resultado por fecha ascendente.

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
df_historial = pd.read_sql(
 """
    SELECT
        duenos.nombre AS nombre_dueno,
        mascotas.nombre AS nombre_mascota,
        consultas.fecha,
        consultas.motivo,
        consultas.costo
    FROM consultas
    JOIN mascotas
        ON consultas.mascota_id = mascotas.id
    JOIN duenos
        ON mascotas.dueno_id = duenos.id
    ORDER BY consultas.fecha ASC
    """,
    conexion
)
print(df_historial)
conexion.close()
```

### Ejercicio 9 — Investigación: alias en SQL
 
Cuando unes varias tablas, los nombres de las columnas pueden colisionar — dos tablas pueden tener una columna que se llama `id`, por ejemplo. Los alias resuelven ese problema.
 
**Consigna:** Investiga cómo usar `AS` en SQL para renombrar columnas y tablas en una consulta. Después reescribí la consulta del Ejercicio 8 usando alias para:
 
1. Renombrar `duenos.nombre` como `dueno` y `mascotas.nombre` como `mascota` en el resultado
2. Usar alias cortos para las tablas dentro de la consulta (por ejemplo, `c` para `consultas`, `m` para `mascotas`, `d` para `duenos`)

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
df_historial = pd.read_sql(
    """
    SELECT
        d.nombre AS dueno,
        m.nombre AS mascota,
        c.fecha AS fecha,
        c.motivo AS motivo,
        c.costo AS costo
    FROM consultas AS c
    JOIN mascotas AS m
        ON c.mascota_id = m.id
    JOIN duenos AS d
        ON m.dueno_id = d.id
    ORDER BY c.fecha ASC
    """,
    conexion
)
 
print(df_historial.columns.tolist())
# Esperado: ['dueno', 'mascota', 'fecha', 'motivo', 'costo']
```

### Ejercicio 10 — Filtrar después de unir

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
df_sosa = pd.read_sql(
   """
    SELECT
        m.nombre AS mascota,
        c.fecha,
        c.motivo,
        c.costo
    FROM consultas AS c
    JOIN mascotas AS m
        ON c.mascota_id = m.id
    WHERE c.veterinario = 'Dra. Sosa'
    ORDER BY c.fecha DESC
    """,
    conexion
)
 
print(df_sosa)
print(f"\nTotal facturado por Dra. Sosa: ${df_sosa['costo'].sum():,.0f}")
```

### Ejercicio 11 — El código que no cierra la conexión
 
Este código funciona, pero tiene un problema que en un script corto no se nota y en uno largo puede causar errores difíciles de rastrear.
 
**Consigna:** Ejecuta este código, identifica el problema, y reescribilo de forma correcta:

```python
def obtener_mascotas(especie):
    conexion = sqlite3.connect("vetcare.db")

    try:
        df = pd.read_sql(
            "SELECT * FROM mascotas WHERE especie = ?",
            conexion,
            params=(especie,)
        )
        return df
    finally:
        conexion.close()

perros = obtener_mascotas("perro")
gatos = obtener_mascotas("gato")
loros = obtener_mascotas("loro")

print(perros)
print(gatos)
print(loros)

```

Hay dos problemas en este código. Uno ya lo mencionamos en el artículo. El otro es de seguridad — investigá qué es una **inyección SQL** y por qué el código de arriba es vulnerable.
 
> **Pista para el segundo problema:** busca "SQL injection" y "parámetros en sqlite3 Python". La solución involucra reemplazar el f-string por algo diferente.

 Hay dos problemas: el primero la conexión no se cierra, y el problema 2 es que la consulta usa un f-string para insertar especie en SQL. eso es lo que debemos corregir. Se cmbió: try: ejecuta la consulta. finally: cierra la conexión siempre, aunque la consulta falle o se produzca un error.WHERE especie = ?: utiliza un marcador de parámetro. y params=(especie,): pasa el valor de forma segura. La coma es necesaria para que Python lo interprete como una tupla de un solo elemento.

### Ejercicio 12 — Consultas condicionales con Python
 
En un sistema real, las consultas que hacés no siempre son las mismas — cambian según lo que el usuario pide o el análisis que estás haciendo. Necesitás construir consultas dinámicamente desde Python.
 
**Consigna:** Escribe una función `buscar_mascotas(especie=None, edad_maxima=None, ciudad=None)` que construya y ejecute una consulta SQL según los parámetros que reciba. Si un parámetro es `None`, no se incluye en el filtro. La función debe devolver un DataFrame.

```python
import sqlite3
import pandas as pd
 
def buscar_mascotas(especie=None, edad_maxima=None, ciudad=None):
    conexion = sqlite3.connect("vetcare.db")
    
    # Tu tarea: construir la consulta dinámicamente
    # Si especie='perro', filtrar por especie
    # Si edad_maxima=5, filtrar mascotas con edad <= 5
    # Si ciudad='Buenos Aires', hacer JOIN con duenos y filtrar por ciudad
    # Si se pasan varios, combinar todos los filtros con AND
    consulta = """
        SELECT m.*
        FROM mascotas AS m
        JOIN duenos AS d ON m.dueno_id = d.id
    """

    filtros = []
    parametros = []

    if especie is not None:
        filtros.append("m.especie = ?")
        parametros.append(especie)

    if edad_maxima is not None:
        filtros.append("m.edad <= ?")
        parametros.append(edad_maxima)

    if ciudad is not None:
        filtros.append("d.ciudad = ?")
        parametros.append(ciudad)

    if filtros:
        consulta += " WHERE " + " AND ".join(filtros)

    try:
        df = pd.read_sql(consulta, conexion, params=parametros)
        return df
    finally:
        conexion.close()

    
    pass
 
# Probar con distintas combinaciones:
print(buscar_mascotas(especie="perro"))
print(buscar_mascotas(edad_maxima=3))
print(buscar_mascotas(especie="gato", ciudad="Buenos Aires"))
print(buscar_mascotas())  # sin filtros → devuelve todas
```

### Ejercicio 13 — Reporte por veterinario
 
La clínica quiere saber cuánto facturó cada veterinario y cuántas consultas atendió. En la próxima clase vas a aprender `GROUP BY` y funciones de agregación en SQL — por ahora, hacelo con las herramientas que ya conocés.
 
**Consigna:** Trae todas las consultas con `pd.read_sql()` y calculá manualmente, usando Python puro (sin `.groupby()` ni `.mean()`), cuántas consultas atendió cada veterinario y cuánto facturó en total. Generá un reporte formateado.

```python
import sqlite3
import pandas as pd

conexion = sqlite3.connect("vetcare.db")
df = pd.read_sql("SELECT veterinario, costo FROM consultas", conexion)
conexion.close()

veterinarios = df["veterinario"].unique()

for nombre in veterinarios:
    consultas_vet = df[df["veterinario"] == nombre]
    cantidad = 0
    total = 0

    for costo in consultas_vet["costo"]:
        cantidad += 1
        total += costo

    print(f"Veterinario: {nombre}")
    print(f"Cantidad de consultas: {cantidad}")
    print(f"Total facturado: ${total:,.0f}")
    print("-" * 35)
    
# Tu tarea: sin groupby() ni mean(), calcular por veterinario:
# - cantidad de consultas
# - total facturado
# Pista: podés usar df["veterinario"].unique() para obtener los
# veterinarios únicos, y df[df["veterinario"] == nombre] para filtrar
```
