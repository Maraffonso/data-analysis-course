## Ejercicio 2 — Limpieza: decisiones justificadas

El dataset tiene problemas que van a afectar el análisis si no los tratás antes.

**Consigna:** Aplica las siguientes limpiezas y justifica cada decisión en un comentario:

- Quedate solo con propiedades en venta en USD (filtra por operation = 'sell' y currency = 'USD'). Responde ¿Por qué tiene sentido trabajar solo en USD? Porque el mercado inmobiliario maneja por lo general números en dólares y al estar todo en una misma moneda es mas facil para obtener datos y realizar operaciones matemáticas. 
- Eliminar filas donde el precio o la superficie sean nulos ¿Qué porcentaje del dataset estás perdiendo? Estoy perdiendo el 59.78% de los registros porque no tenían el precio o la superficie informados.
- Filtrar precios que parezcan errores evidentes ¿Tiene sentido una propiedad que vale $1 USD? ¿Y una de $50 millones?.Define un criterio para filtrar y justificalo. Puede que sean errores de tipeo porque no puede valer 1 usd y otra de 50 millones podria ser pero es un valor muy elevado a menos que sea un terreno de muchos m2 para una inversión o algo de eso. Por eso limpiarlos datos paso a paso sirve exactamente para eso: dejar solo la información coherente, lógica y útil para empezar a sacar conclusiones de verdad.
- Quedate solo con los tipos de propiedad principales (por ejemplo: Departamento, Casa, PH) ¿Por qué descartas los demás? Me quedaron tras conservar solo departamentos, casas y PHs: filas  215287 descarte las demas. Descartamos lo demas que no sea para habitar como galpones, cocheras o los locales comerciales.El objetivo es analizar el mercado residencial (de tipo vivienda), mezclar estos otros tipos de propiedades arruinaría las estadísticas y crearía conclusiones falsas como dice el dicho sería como comparar peras con manzanas.
- Al terminar: ¿cuántas filas quedaron? ¿Qué porcentaje del original?
Me quedaron 215,287 filas limpias y teniendo en cuenta que tenia 992,192 registros inicialmente se conservó un 27,70% de los datos, 78,30% se descartó por su información ej precios usd 1. 

````python
# filtro por Venta y USD

col_op = 'operation_type' if 'operation_type' in df.columns else 'operation'
df_venta_usd = df[(df[col_op] == 'Venta') & (df['currency'] == 'USD')].copy()
filas_antes_nulos = len(df_venta_usd)

````

````python
# 2 Elimino los nulos en precio y superficie total
df_sin_nulos = df_venta_usd.dropna(subset=['precio', 'surface_total']).copy()
filas_despues_nulos = len(df_sin_nulos)
````

````python
porcentaje_perdido = ((filas_antes_nulos - filas_despues_nulos) / filas_antes_nulos) * 100

(f"Filas antes de eliminar nulos: {filas_antes_nulos}")
print(f"Filas después de eliminar nulos: {filas_despues_nulos}")
print(f"Porcentaje de datos perdidos por nulos: {porcentaje_perdido:.2f}%")
````

````python
# punto 3
filas_antes_outliers = len(df_sin_nulos)

df_filtrado_outliers = df_sin_nulos[
    (df_sin_nulos['precio'] >= 10000) & (df_sin_nulos['precio'] <= 2500000) &
    (df_sin_nulos['surface_total'] >= 10) & (df_sin_nulos['surface_total'] <= 1000)
].copy()

filas_despues_outliers = len(df_filtrado_outliers)
print(f"Filas antes de filtrar outliers: {filas_antes_outliers}")
print(f"Filas después de filtrar outliers: {filas_despues_outliers}")
````

````python
# punto 4 tipos de prop
tipos_principales = ['Departamento', 'Casa', 'PH']
df_filtrado_tipos = df_filtrado_outliers[df_filtrado_outliers['tipo_propiedad'].isin(tipos_principales)].copy()

print(f"Filas tras conservar solo departamentos, casas y PHs: {len(df_filtrado_tipos)}")
````

````python
#punto 5 datos finales y que porcentaje de la info nos quedo tras la limpiesa 
filas_iniciales = len(df)
filas_finales = len(df_filtrado_tipos)

porcentaje_conservado = (filas_finales / filas_iniciales) * 100
porcentaje_perdido = 100 - porcentaje_conservado

print(f"--- REPORTE FINAL DE LIMPIEZA ---")
print(f"Filas iniciales: {filas_iniciales:,}")
print(f"Filas finales (limpias): {filas_finales:,}")
print(f"Porcentaje de datos conservados: {porcentaje_conservado:.2f}%")
print(f"Porcentaje total descartado: {porcentaje_perdido:.2f}%")
````
